# Tell Me Why

> Bibliothèque nbdev pour entraîner, greffer et analyser un adversarial autoencoder orienté explicabilité.

`Tell Me Why` reconstruit progressivement les briques utiles de la branche `Arda` de `Human-Centered-xAI` sous forme de librairie Python maintenable. Le code source reste dans les notebooks `nbs/`, puis nbdev génère le package `tell_me_why/`.

## Workflow nbdev

1. Le code maintenu vit dans les notebooks `nbs/*.ipynb`.
2. Les cellules exportées vers la librairie portent la directive `#| export`.
3. La commande `nbdev_export` régénère les modules Python dans `tell_me_why/`.
4. Les tests vérifient que le package importe correctement et que les briques stabilisées restent utilisables.

Les fichiers `.py` générés ne sont pas la source de vérité: toute modification durable doit être faite dans les notebooks.

## Carte des notebooks

- `00_init.ipynb`: version et initialisation minimale du package.
- `01_model_aae.ipynb`: définition du modèle `AAE`, du décodeur U-Net avec dropout sur les skip connections, et des fonctions de perte associées.
- `02_feature_scores.ipynb`: fonctions de calcul des scores de features à réutiliser dans les analyses.
- `03_graft_classifier.ipynb`: logique pour remplacer ou brancher une tête de classification externe sur le modèle AAE.
- `04_visualization.ipynb`: visualisations de l'espace latent et des scores de features.
- `05_tutorial.ipynb`: exemple complet d'utilisation de la librairie, sans obligation d'exporter du code.

## Priorité de développement

Le premier objectif est de stabiliser une API minimale autour du modèle AAE, puis d'ajouter les briques d'analyse une par une.

1. Finaliser et exporter `01_model_aae.ipynb`.
2. Ajouter les scores réutilisables dans `02_feature_scores.ipynb`.
3. Définir une interface simple de greffe dans `03_graft_classifier.ipynb`.
4. Brancher les visualisations dans `04_visualization.ipynb`.
5. Transformer `05_tutorial.ipynb` en parcours utilisateur quand les modules précédents sont utilisables.
