# Tell Me Why

> Greffer un classifieur fastai/PyTorch sur `modelAAE_DROPOUT.py` depuis la branche `Arda` de `LucaLaFisca/Human-Centered-xAI` et inspecter son espace latent.

Ce dépôt est configuré pour nbdev 3. Les notebooks `nbs/00_source.ipynb` à `nbs/08_plots.ipynb` sont les sources exportables des modules Python, et `nbs/09_tutorial.ipynb` sert d'exemple d'utilisation.

Règle importante: on modifie les notebooks dans `nbs/`, puis on lance `nbdev-export`. Les fichiers `tell_me_why/*.py` sont générés par nbdev et ne doivent pas être édités directement.

La structure suit une logique proche de FasterAI: un module court par responsabilité, puis une API simple ré-exportée dans `tell_me_why`.

Installation développeur:

```bash
conda env create -f environment.yml
conda activate tell-me-why
python -m pip install -e ".[dev]"
```

Cycle nbdev courant:

```bash
nbdev-export
nbdev-prepare
nbdev-test
```
